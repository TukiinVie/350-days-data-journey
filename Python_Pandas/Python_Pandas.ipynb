```python
import pandas as pd

#1. Khởi taoh bảng dữ liệu mô phỏng giao dịch ngân hàng

data = {
    'txn_id' : ['T01', 'T02', 'T03', 'T04', 'T05'],
    'customer_id' : ['C01', 'C02', 'C03', 'C04', 'C05'],
    'branch': ['Hanoi', 'HCM', 'Hanoi', 'Danang', 'HCM'],
    'amount': [5000000, 12000000, 2000000, 8000000, 250000000],
    'status' : ['SUCESS', 'SUCCESS', 'SUCCESS', 'FAILED', 'SUCCESS']
}

df = pd.DataFrame(data)

#2. Khám phá nhanh 5 câu lệnh quan trọng
print("--1. Xem 3 dòng đầu --")
print(df.head(3))

print("\n--- 2. Kích thước bảng (Số dòng, Số cột) ---")
print(df.shape)

print("\n--- 3. Xem cấu trúc dữ liệu và kiểu cột ---")
df.info()

print("\n--- 4. Thống kê nhanh các chỉ số số học (Mean, Min, Max) ---")
print(df.describe())

print("\n--- 5. Đếm số lượng giao dịch theo từng chi nhánh ---")
print(df['branch'].value_counts())
```
